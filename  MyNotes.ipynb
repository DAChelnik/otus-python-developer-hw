Моё видение исполнения кода
=====
> 23 сентября 2026 года

После запуска на экране появляется меню из восьми пунктов. Ввожу нужный номер и нажимаю Enter:

- *"Открыть файл"* - программа попросит указать, где лежит справочник. Можно ввести путь к своему файлу, а можно просто нажать Enter. В этом случае будет взят стандартный файл phonebook.json. Если он уже существует, из него загрузятся все записи. Если файла нет, программа создаст пустой справочник и будет ждать, пока я начну его наполнять. Пока справочник не открыт, остальные действия недоступны, искать или удалять контакты в пустом списке программа не даст.


- *"Сохранить файл"* - все добавленные и изменённые записи будут записаны на диск. Здесь тоже можно либо указать свой путь, либо сохранить данные в тот же файл, из которого был открыт справочник.

- *"Показать все контакты"* - программа выведет на экран весь список. Если список пуст, она должна сообщитьоб этом.

- *"Создать контакт"* - программа по очереди попросит ввести данные контакта и адрес. Контакту присовится свой уникальный номер ID. В идеале бы хранить адреса в отлеьном json и добавить проверку на уникальность фио и телефона. и телефон вводить в едином стандарте.

- *"Найти контакт"* - программа спросит по какому полю искать. После ввода покажет все контакты, где оно встречается. Регистр не важен, поэтому «иван» и «Иван» для неё одинаковые. 

- *"Изменить контакт"* - ввожу ID нужной записи, после чего программа показывает её данные и предлагает заново заполнить каждое поле. Если какое-то значение трогать не нужно, просто нажать Enter и старый вариант сохранится.

- *"Удалить контакт"* - ввожу ID нужной записи, программа показывает сам контакт и спрашивает, точно ли его нужно удалить. Если подтвердить это, введя "y", запись исчезнет из списка.

- *"Выход"* - давай до свидания, и закрываемся. Но, сначала нужно проверить, есть ли данные которые менялись, но не сохарнились.

На заметку
---
> contacts: list[dict[str, Any]
`contacts`: это весь список (тип list)
ожидается список словарей. каждый словарь — контакт. 
dict[str, Any] — тип «словарь», где:
- ключи — строки (str: например, "name", "phone"),
- значения — любого типа (Any: число, строка, список, другой словарь и т. д.).

Поля для контакта
---

ID: id
Имя: name
Телефон: phone

Комментарий: comment

Страна: country
Область: region
Город: city
Улица: street
Дом: house
Здание: building
Организация: organization

На заметку
---

Вылезла ошибка:
> Неразрешённая ссылка 'Any'
для решения добавил строку импорта 
> from typing import Any
`typing` - это модуль для аннотаций типов
`Any` означает «значение любого типа», используется когда мы заранее не знаем, какой тип данных окажется в переменной.

Примерный сценарий
----
Представим, что пользователь запускает программу впервые. Файла phonebook.json ещё нет. Затем он:

Открывает файл (файла нет → пустой справочник)
Создаёт контакт Анны
Создаёт контакт Бориса
Пытается создать дубликат Анны
Показывает все контакты
Ищет по имени «анна»
Ищет по всем полям «краснодар»
Редактирует контакт Анны
Удаляет контакт Бориса
Сохраняет файл
Выходит без несохранённых изменений
Снова открывает файл, меняет комментарий, выходит с несохранёнными изменениями
